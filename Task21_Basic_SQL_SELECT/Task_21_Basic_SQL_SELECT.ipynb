**## Import libraries**

```python
import sqlite3
import pandas as pd
from IPython.display import display

print("Libraries imported successfully!")
```

**## Read datasets**

```python
customers = pd.read_csv("customers.csv")
products = pd.read_csv("products.csv")

print("Customers loaded:", len(customers))
print("Products loaded:", len(products))

print("\nCustomers columns:")
print(customers.columns.tolist())

print("\nProducts columns:")
print(products.columns.tolist())
```

**## Create database**

```python
conn = sqlite3.connect(":memory:")

customers.to_sql(
    "Customers", conn, index=False, if_exists="replace"
)

products.to_sql(
    "Products", conn, index=False, if_exists="replace"
)

print("Database created successfully!")
```

**## Query 1: Display all customers**

```python
query = """
SELECT *
FROM Customers;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

**## Query 2: Select customer details**

```python
query = """
SELECT CustomerID, CompanyName, City, Country
FROM Customers;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

**## Query 3: Customers from London**

```python
query = """
SELECT CustomerID, CompanyName, City
FROM Customers
WHERE City = 'London';
"""

result = pd.read_sql_query(query, conn)
display(result)
```

**## Query 4: Customers from Germany**

```python
query = """
SELECT CustomerID, CompanyName, Country
FROM Customers
WHERE Country = 'Germany';
"""

result = pd.read_sql_query(query, conn)
display(result)
```

**## Query 5: Sort customers by company name**

```python
query = """
SELECT CustomerID, CompanyName, Country
FROM Customers
ORDER BY CompanyName ASC;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

**## Query 5: Sort customers by company name**

```python
query = """
SELECT *
FROM Products;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

**## Query 6: Display all products**

```python
query = """
SELECT ProductID, ProductName, UnitPrice
FROM Products
WHERE UnitPrice < 20;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

**## Query 7: Products priced below 20**

```python
query = """
SELECT ProductID, ProductName, UnitPrice
FROM Products
WHERE UnitPrice > 20;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

**## Query 8: Products priced above 20**

```python
query = """
SELECT ProductID, ProductName, UnitPrice
FROM Products
ORDER BY UnitPrice DESC;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

**## Query 9: Sort products by highest price**

```python
query = """
SELECT ProductID, ProductName, UnitPrice
FROM Products
ORDER BY UnitPrice ASC
LIMIT 5;
"""

result = pd.read_sql_query(query, conn)
display(result)
```
