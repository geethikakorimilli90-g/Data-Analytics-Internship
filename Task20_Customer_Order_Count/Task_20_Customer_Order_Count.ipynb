**## Upload Dataset**

```python
from google.colab import files
import pandas as pd

uploaded = files.upload()

file_name = next(iter(uploaded))
df = pd.read_excel(file_name)
```

```python
print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)
display(df.head())
print("Column Names:", df.columns.tolist())
```

**## Load Dataset into SQL**

```python
import sqlite3

conn = sqlite3.connect(":memory:")

df.columns = df.columns.str.strip()

df.to_sql(
    "OnlineRetail",
    conn,
    index=False,
    if_exists="replace"
)

print("Dataset connected to SQLite successfully!")
```

**## Check Missing Customer IDs**

```python
query = """
SELECT
    COUNT(*) AS TotalRows,
    SUM(CASE WHEN CustomerID IS NULL THEN 1 ELSE 0 END)
        AS MissingCustomerRows
FROM OnlineRetail;
"""

display(pd.read_sql_query(query, conn))
```

**## Customer-wise Order Count**

```python
query = """
SELECT
    CustomerID,
    COUNT(DISTINCT InvoiceNo) AS OrderCount
FROM OnlineRetail
WHERE CustomerID IS NOT NULL
GROUP BY CustomerID
ORDER BY OrderCount DESC;
"""

result = pd.read_sql_query(query, conn)

display(result.head(20))

print("Total Customers:", len(result))
```

```python
from google.colab import files

sql_query_content = """SELECT
    CustomerID,
    COUNT(DISTINCT InvoiceNo) AS OrderCount
FROM OnlineRetail
WHERE CustomerID IS NOT NULL
GROUP BY CustomerID
ORDER BY OrderCount DESC;"""

# Save the query content into a .sql file
with open("customer_order_count.sql", "w") as f:
    f.write(sql_query_content)

# Download the file to your local computer
files.download("customer_order_count.sql")
```

**## Top 10 Customers**

```python
result.to_csv("customer_order_count.csv", index=False)
top_10.to_csv("top_10_customers.csv", index=False)

print("CSV files saved successfully!")
```
