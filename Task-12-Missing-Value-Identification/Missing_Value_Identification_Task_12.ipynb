
**## Load the Dataset**

```python
import pandas as pd

df = pd.read_csv('/content/cleaned_titanic.csv')

df.head()
```

**## Identify Missing Values**

```python
# Check total nulls and percentage
missing_count = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df)) * 100

summary = pd.DataFrame({
    'Missing Count': missing_count,
    'Percentage (%)': missing_percent.round(2)
})

print("--- MISSING VALUE AUDIT REPORT ---")
print(summary)
```

**## Summary and Findings**

```python

# Short Findings
total_missing = df.isnull().sum().sum()
columns_with_missing = df.columns[df.isnull().any()].tolist()

print("SHORT FINDINGS")
print("Total missing values:", total_missing)
print("Columns with missing values:", columns_with_missing)

if total_missing == 0:
    print("No missing values were found in this dataset.")
else:
    print("Missing values were found in the dataset.")
    print("Review the missing-value count and percentage for each column.")
```
