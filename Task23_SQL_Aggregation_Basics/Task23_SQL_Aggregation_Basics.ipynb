**## Import libraries**

```python
import pandas as pd
import sqlite3
from IPython.display import display
```

**## Load the dataset**

```python
df = pd.read_csv('/content/employees_dataset.csv')

print("Dataset loaded successfully!")
display(df)
```

**## Check dataset information**

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())
```

**## Create SQL database and table**

```python
conn = sqlite3.connect(':memory:')

df.to_sql('employees', conn, if_exists='replace', index=False)

print("SQL table created successfully!")
```

**## Display all employees**

```python
query = """
SELECT *
FROM employees;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

**## COUNT(): Total number of employees**

```python
query = """
SELECT COUNT(*) AS total_employees
FROM employees;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

**## SUM(): Total salary**

```python
query = """
SELECT SUM(salary) AS total_salary
FROM employees;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

**## AVG(): Average salary**

```python
query = """
SELECT ROUND(AVG(salary), 2) AS average_salary
FROM employees;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

**## MIN(): Minimum salary**

```python
query = """
SELECT MIN(salary) AS minimum_salary
FROM employees;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

**## MAX(): Maximum salary**

```python
query = """
SELECT MAX(salary) AS maximum_salary
FROM employees;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

**## GROUP BY: Employee count by department**

```python
query = """
SELECT department, COUNT(*) AS employee_count
FROM employees
GROUP BY department
ORDER BY department;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

**## GROUP BY: Employee count by department**

```python
query = """
SELECT department, SUM(salary) AS total_salary
FROM employees
GROUP BY department
ORDER BY department;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

**## GROUP BY: Average salary by department**

```python
query = """
SELECT department, ROUND(AVG(salary), 2) AS average_salary
FROM employees
GROUP BY department
ORDER BY department;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

**## MIN() and MAX() by department**

```python
query = """
SELECT
    department,
    MIN(salary) AS minimum_salary,
    MAX(salary) AS maximum_salary
FROM employees
GROUP BY department
ORDER BY department;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

**##HAVING: Departments with average salary above 45000**

```python
query = """
SELECT
    department,
    ROUND(AVG(salary), 2) AS average_salary
FROM employees
GROUP BY department
HAVING AVG(salary) > 45000
ORDER BY department;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

**## ORDER BY: Employees ranked by salary**

```python
query = """
SELECT name, department, salary
FROM employees
ORDER BY salary DESC;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

**## Transaction and ROLLBACK**

```python
cursor = conn.cursor()

cursor.execute("BEGIN TRANSACTION")

cursor.execute("""
INSERT INTO employees
(employee_id, name, department, salary, age)
VALUES (11, 'Temporary', 'IT', 30000, 22);
""")

print("After INSERT:")
print(pd.read_sql_query(
    "SELECT COUNT(*) AS total_employees FROM employees;",
    conn
))

conn.rollback()

print("\nAfter ROLLBACK:")
print(pd.read_sql_query(
    "SELECT COUNT(*) AS total_employees FROM employees;",
    conn
))
```

**## Final summary of aggregate functions**

```python
query = """
SELECT
    COUNT(*) AS total_employees,
    SUM(salary) AS total_salary,
    ROUND(AVG(salary), 2) AS average_salary,
    MIN(salary) AS minimum_salary,
    MAX(salary) AS maximum_salary
FROM employees;
"""

result = pd.read_sql_query(query, conn)
display(result)
```
