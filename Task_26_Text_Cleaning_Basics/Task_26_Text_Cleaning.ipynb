**## Text Cleaning Basics - Customer Data Cleaning**

```python

import pandas as pd


df = pd.read_excel("/content/Task-26-Text-Cleaning-Dataset.xlsx")


cleaned_df = df.copy()


for col in ["Customer_Name", "City"]:
    cleaned_df[col] = (
        cleaned_df[col]
        .fillna("")
        .astype(str)
        .str.replace("\u00A0", " ", regex=False)
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
        .str.title()
    )

# Show original data
print("BEFORE CLEANING")
display(df)

# Show cleaned data
print("AFTER CLEANING")
display(cleaned_df)

# Show before-and-after examples
comparison = pd.DataFrame({
    "Before_Name": df["Customer_Name"],
    "After_Name": cleaned_df["Customer_Name"],
    "Before_City": df["City"],
    "After_City": cleaned_df["City"]
})

print("BEFORE AND AFTER COMPARISON")
display(comparison)

# Save the final files
cleaned_df.to_csv("cleaned_customer_dataset.csv", index=False)
comparison.to_csv("before_after_examples.csv", index=False)

cleaned_df.to_excel("Task-26-Cleaned-Dataset.xlsx", index=False)

print("SUCCESS: All output files created!")
```
