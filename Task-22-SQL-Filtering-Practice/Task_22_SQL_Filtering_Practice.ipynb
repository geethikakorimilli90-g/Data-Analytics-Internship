```python

import os

os.makedirs("Task_22_Results", exist_ok=True)

queries = {
    "Query_01": query1,
    "Query_02": query2,
    "Query_03": query3,
    "Query_04": query4,
    "Query_05": query5,
    "Query_06": query6,
    "Query_07": query7,
    "Query_08": query8,
    "Query_09": query9,
    "Query_10": query10,
    "Query_11": query11,
    "Query_12": query12
}

for name, sql in queries.items():
    result = pd.read_sql_query(sql, conn)
    result.to_csv(
        f"Task_22_Results/{name}_Results.csv",
        index=False
    )

with open("Task_22_Results/SQL_Queries.txt", "w") as f:
    for name, sql in queries.items():
        f.write(f"{name}\n")
        f.write(sql.strip() + ";\n\n")

print("All 12 query results saved successfully!")
```
